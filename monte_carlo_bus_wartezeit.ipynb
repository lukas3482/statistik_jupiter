# Monte-Carlo-Simulation: Wartezeit auf den Bus

## Aufgabe

Eine Person fährt jeden Morgen mit dem Bus zur FH.  
Der Bus fährt laut Fahrplan alle **15 Minuten**.

Da die Person zufällig an der Haltestelle ankommt, kann die Wartezeit zwischen **0 und 15 Minuten** liegen.

Zusätzlich hat der Bus manchmal Verspätung:

- mit einer Wahrscheinlichkeit von **80 %** ist der Bus pünktlich,
- mit einer Wahrscheinlichkeit von **15 %** hat der Bus zwischen **1 und 5 Minuten** Verspätung,
- mit einer Wahrscheinlichkeit von **5 %** hat der Bus zwischen **5 und 15 Minuten** Verspätung.

Es werden **100.000 Morgen** simuliert.

## Fragestellung

1. Wie hoch ist die Wahrscheinlichkeit, dass die gesamte Wartezeit mehr als **15 Minuten** beträgt?
2. Wie groß ist die durchschnittliche Wartezeit?
3. Wie stark schwanken die Wartezeiten?
4. Stellen Sie die simulierte Verteilung der Wartezeit grafisch dar.


## Idee der Monte-Carlo-Simulation

Für jeden simulierten Morgen wird zufällig bestimmt:

1. wann die Person relativ zum Fahrplan ankommt,
2. ob der Bus pünktlich ist oder Verspätung hat,
3. wie groß die gesamte Wartezeit ist.

Die gesamte Wartezeit ergibt sich aus:

\[
\text{Wartezeit} = \text{zufällige Wartezeit bis zum geplanten Bus} + \text{Verspätung}
\]


```python
import numpy as np
import matplotlib.pyplot as plt

# Reproduzierbarkeit
np.random.seed(42)

# Anzahl der Simulationen
n = 100_000

# Zufällige Wartezeit bis zum geplanten Bus: gleichverteilt zwischen 0 und 15 Minuten
wartezeit_fahrplan = np.random.uniform(0, 15, n)

# Zufallszahlen zur Entscheidung, ob der Bus pünktlich ist oder Verspätung hat
zufall = np.random.random(n)

# Verspätung initialisieren
verspaetung = np.zeros(n)

# 15 %: kleine Verspätung zwischen 1 und 5 Minuten
kleine_verspaetung = (zufall >= 0.80) & (zufall < 0.95)
verspaetung[kleine_verspaetung] = np.random.uniform(1, 5, kleine_verspaetung.sum())

# 5 %: größere Verspätung zwischen 5 und 15 Minuten
grosse_verspaetung = zufall >= 0.95
verspaetung[grosse_verspaetung] = np.random.uniform(5, 15, grosse_verspaetung.sum())

# Gesamte Wartezeit
wartezeit_gesamt = wartezeit_fahrplan + verspaetung

# Erste 10 simulierte Wartezeiten anzeigen
wartezeit_gesamt[:10]
```

## Wahrscheinlichkeit für mehr als 15 Minuten Wartezeit

Nun wird berechnet, wie oft die gesamte Wartezeit größer als 15 Minuten ist.


```python
wahrscheinlichkeit_mehr_als_15 = np.mean(wartezeit_gesamt > 15)

print(f"Wahrscheinlichkeit für mehr als 15 Minuten Wartezeit: {wahrscheinlichkeit_mehr_als_15:.4f}")
print(f"Das entspricht ca. {wahrscheinlichkeit_mehr_als_15 * 100:.2f} %")
```

## Mittelwert und Standardabweichung

Der Mittelwert beschreibt die durchschnittliche Wartezeit.  
Die Standardabweichung beschreibt, wie stark die Wartezeiten typischerweise um den Mittelwert schwanken.


```python
mittelwert = np.mean(wartezeit_gesamt)
standardabweichung = np.std(wartezeit_gesamt)

print(f"Mittlere Wartezeit: {mittelwert:.2f} Minuten")
print(f"Standardabweichung: {standardabweichung:.2f} Minuten")
```

## Interpretation

Die simulierte mittlere Wartezeit liegt ungefähr bei **8,45 Minuten**.

Ohne Verspätung wäre die durchschnittliche Wartezeit bei einem 15-Minuten-Takt:

\[
\frac{15}{2} = 7{,}5
\]

Durch die zufälligen Verspätungen steigt die durchschnittliche Wartezeit also an.

Die Standardabweichung liegt ungefähr bei **4,8 Minuten**.  
Das bedeutet, dass die Wartezeit von Tag zu Tag deutlich schwanken kann.

Die Wahrscheinlichkeit, mehr als 15 Minuten warten zu müssen, liegt ungefähr bei **6 %**.


## Visualisierung

Das Histogramm zeigt die Verteilung der simulierten gesamten Wartezeit.


```python
plt.figure(figsize=(9, 5))

plt.hist(wartezeit_gesamt, bins=40, edgecolor="black")

plt.title("Monte-Carlo-Simulation der Bus-Wartezeit")
plt.xlabel("Gesamte Wartezeit in Minuten")
plt.ylabel("Häufigkeit")
plt.grid(axis="y", alpha=0.3)

plt.show()
```

## Kurzlösung

Bei den gewählten Annahmen ergibt die Monte-Carlo-Simulation ungefähr:

\[
P(X > 15) \approx 6 \%
\]

\[
E(X) \approx 8{,}45 \text{ Minuten}
\]

\[
\sigma(X) \approx 4{,}8 \text{ Minuten}
\]

Die Person wartet also im Durchschnitt etwas mehr als 8 Minuten.  
An etwa 6 von 100 Tagen wartet sie länger als 15 Minuten.
