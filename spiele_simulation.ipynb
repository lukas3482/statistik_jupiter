```python
import numpy as np
import matplotlib.pyplot as plt

# 0 -> Zahl
# 1 -> Kopf
n = 10**5
data = np.random.choice(np.arange(0,2), size=(n, 10))
gesamtGewinn = 0

for i in range(n):
    gewinn = 0
    for j in range(10):
        if data[i][j] == 1:
            gewinn += 3
    gesamtGewinn += gewinn

print(f"Durchschnittsgewinn: {gesamtGewinn/n}")
print(f"Gesamt: {gesamtGewinn}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 0 -> Zahl
# 1 -> Kopf
n = 10**5
data = np.random.choice(np.arange(0,2), size=(n, 10))
gesamtGewinn = 0

for i in range(n):
    gewinn = 10
    for j in range(10):
        if data[i][j] == 1:
            gewinn *= 1.2
        else:
            gewinn *= 0.8

    gesamtGewinn += gewinn

print(f"Durchschnittsgewinn: {gesamtGewinn/n}")
```

```python
import numpy as np
import random
import matplotlib.pyplot as plt

# 0 -> Zahl
# 1 -> Kopf
n = 10**5
gesamtGewinnArr = []

for i in range(n):
    gewinn = 1
    while True:
        if random.randint(0, 1) == 1:
            gewinn *= 2
        else:
            break
    gesamtGewinnArr.append(gewinn)

print(f"Durchschnittsgewinn: {np.mean(gesamtGewinnArr)}")
print(f"Gesamt: {np.sum(gesamtGewinnArr)}")
print(f"Median: {np.median(gesamtGewinnArr)}")
print(f"Max: {np.max(gesamtGewinnArr)}")
```
